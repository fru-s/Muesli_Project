```python
orders_clean = orders_raw.copy()                        # making a copy of the database for cleaning purposes
orders_clean.columns = orders_clean.iloc[0]
orders_clean = orders_clean.iloc[1:].reset_index(drop=True)
orders_clean

```

```python
orders_clean.columns.tolist()     # checking column names, we will need to change it to snake_case
```

```python
orders_clean.columns = (orders_clean.columns
                         .str.strip()
                         .str.replace('/', '_', regex=False)                   # turn / into _ first
                         .str.replace('-', '_', regex=False)                   # - → _
                         .str.replace(r'[^a-zA-Z0-9\s_]', '', regex=True)      # then delete anything else unwanted
                         .str.replace(r'\s+', '_', regex=True)
                         .str.lower())
orders_clean.columns.tolist()
```

```python
# as a business one of the most interesting kpi is always the profit
import matplotlib.pyplot as plt

orders_clean['profit'].plot.hist(bins=50)
plt.show()
# the histogram looks weird, it refers back to the earlier seen extreme minimum and maximum profit
```

```python
lowest_10 = orders_clean.sort_values('profit').head(10)
lowest_10
# -6599.9780 is not the only extreme high profit, we have several others
# at first glance it doesn't seem that there is a correlation between higher the sales, higher the loss
# but thre seems to a correlation between higher discount leading to higher loss
```

```python
highest_10 = orders_clean.sort_values('profit', ascending=False).head(10)
highest_10
# orders with the highest profit seem to have high sales and no discount at the first glance
```

```python
# want to calculate a cost column
# to double check the hypotheses that higher discount leads to lower profit, lets plot the data
orders_clean.plot.scatter(x='discount', y='profit', alpha=0.3)
plt.axhline(0, color='red', linestyle='--', label='Break-even')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.title('Discount vs. Profit')
plt.legend()
plt.show()
# vertical stripes represent fixed discount tier patterns 
# At discount = 0: almost all points sit above the break-even line — profit is generally positive, with some outliers reaching very high profit (up to 8000)
# As discount increases (moving right): the cluster of points visibly shifts downward, more and more points fall below the red break-even line
# By discount = 0.7–0.8: nearly every point is below zero — meaning at high discounts, orders are consistently unprofitable

```
