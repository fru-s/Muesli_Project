```python
orders_clean = orders_raw.copy()                        # making a copy of the database for cleaning purposes
orders_clean.columns = orders_clean.iloc[0]
orders_clean = orders_clean.iloc[1:].reset_index(drop=True)
orders_clean

```

```python
orders_clean.columns.tolist()     # checking column names, we will need to change it to snake_case
```

```python
orders_clean.columns = (orders_clean.columns
                         .str.strip()
                         .str.replace('/', '_', regex=False)                   # turn / into _ first
                         .str.replace('-', '_', regex=False)                   # - → _
                         .str.replace(r'[^a-zA-Z0-9\s_]', '', regex=True)      # then delete anything else unwanted
                         .str.replace(r'\s+', '_', regex=True)
                         .str.lower())
orders_clean.columns.tolist()
```

In order to really understand the data, we need to know what information the different columns contain.    
Here are the explanations of the individual variables:


| Column name | Description |
| --- | ----------- |
| index | Unclear, doesn't appear on other tabs |
| order_id | Unique identifier of the order |
| order_date | Day and time of the placement of the order |
| ship_mode | Mode of the shipping: First Class or Second Class |
| customer_id  | Unique identifier of the customer |
| customer_name | First and last nem of the customer |
| origin_channel | How the customer placed the order or found the business? |
| origin_region | Country of the customer |
| city | City of the customer |
| state | State of the customer |
| postal_code | Postal code of the customer |
| region | Region based on internal criteria |
| category | Product Category |
| sub_category | Product sub-category |
| product_id | Unique identifier of the product |
| sales | Total amount paid by the customer (net or gross) |
| quantity | Number of products order |
| discount | Discount |
| profit | Profit, calculation is unclear |
