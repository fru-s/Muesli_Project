```python
orders_clean.columns = (orders_clean.columns
                         .str.strip()
                         .str.replace('/', '_', regex=False)                   # turn / into _ first
                         .str.replace('-', '_', regex=False)                   # - → _
                         .str.replace(r'[^a-zA-Z0-9\s_]', '', regex=True)      # then delete anything else unwanted
                         .str.replace(r'\s+', '_', regex=True)
                         .str.lower())
orders_clean.columns.tolist()
```

```python
# want to calculate a cost column
# to double check the hypotheses that higher discount leads to lower profit, lets plot the data
orders_clean.plot.scatter(x='discount', y='profit', alpha=0.3)
plt.axhline(0, color='red', linestyle='--', label='Break-even')
plt.xlabel('Discount')
plt.ylabel('Profit')
plt.title('Discount vs. Profit')
plt.legend()
plt.show()
# vertical stripes represent fixed discount tier patterns 
# At discount = 0: almost all points sit above the break-even line — profit is generally positive, with some outliers reaching very high profit (up to 8000)
# As discount increases (moving right): the cluster of points visibly shifts downward, more and more points fall below the red break-even line
# By discount = 0.7–0.8: nearly every point is below zero — meaning at high discounts, orders are consistently unprofitable

```

```python
sample_product = orders_clean[orders_clean['product_id'] == orders_clean['product_id'].iloc[0]]
print(sample_product[['product_id', 'quantity', 'discount', 'sales']].sort_values('discount'))
# At discount=0: 90.570 / 3 = 30.19 per unit — this is the true, undiscounted unit price.
# 7*30.19=211,33 with 20% discount, 169,64
# this means sales already includes the discount
```

```python
orders_clean['sales_before_discount'] = orders_clean['sales'] / (1 - orders_clean['discount'])      # creating a sales_before_discount_column
orders_clean['cost'] = orders_clean['sales'] - orders_clean['profit']                               # cost column
orders_clean.head()
```

```python
order_summary = orders_clean.groupby('order_id').agg(
    total_sales=('sales', 'sum'),
    total_profit=('profit', 'sum'),
    total_sales_before_discount=('sales_before_discount', 'sum'),
    total_cost=('cost', 'sum'),
    avg_discount=('discount', 'mean'),
    total_quantity=('quantity', 'sum'),
    num_products=('product_id', 'count'),
    customer_id=('customer_id', 'first'),
    order_date=('order_date', 'first'),
    city=('city', 'first'),
    state=('state', 'first')
).reset_index()

order_summary.head(10)
# can be used for merging with other tables
```

```python
order_summary.shape
# matches the unique value of the order numbers
```

```python
fig, ax = plt.subplots(2, 3, figsize=(20, 10))
plt.suptitle('Distribution of numeric columns', fontsize=20)
fig.tight_layout()
plt.subplots_adjust(hspace=.5, wspace=.2, top=.9)

ax[0][0].boxplot(x=order_summary['total_sales'])
ax[0][0].set_xticklabels(labels=["Total Sales"])
ax[0][0].set_title("Distribution of Total Sales", fontsize=15)

ax[0][1].boxplot(x=order_summary['total_profit'])
ax[0][1].set_xticklabels(labels=["Total Profit"])
ax[0][1].set_title("Distribution of Total Profit", fontsize=15)

ax[0][2].boxplot(x=order_summary['total_cost'])
ax[0][2].set_xticklabels(labels=["Total Cost"])
ax[0][2].set_title("Distribution of Total Cost", fontsize=15)

ax[1][0].boxplot(x=order_summary['total_quantity'])
ax[1][0].set_xticklabels(labels=["Total Quantity"])
ax[1][0].set_title("Distribution of Total Quantity", fontsize=15)

ax[1][1].boxplot(x=order_summary['avg_discount'])
ax[1][1].set_xticklabels(labels=["Avg Discount"])
ax[1][1].set_title("Distribution of Avg Discount", fontsize=15)

fig.delaxes(ax[1][2])
```

```python
orders_clean['product_id'].value_counts()
# we sold 1862 different type of products in 4 years
```

```python
top_products_by_year = orders_clean.groupby(['year', 'product_id']).size().reset_index(name='order_count')
top_products_by_year = top_products_by_year.sort_values(['year', 'order_count'], ascending=[True, False])
top_5_per_year = top_products_by_year.groupby('year').head(5)
print(top_5_per_year)
# let's see if the most popular proct changed or stayed the same
# they ae completely different every year
```

```python
for year in sorted(orders_clean['year'].unique()):
    year_top = top_5_per_year[top_5_per_year['year'] == year]
    year_top.plot.bar(x='product_id', y='order_count')
    plt.title(f'Top 5 Products — {year}')
    plt.xlabel('Product ID')
    plt.ylabel('Times Ordered')
    plt.xticks(rotation=45)
    plt.show()
```
